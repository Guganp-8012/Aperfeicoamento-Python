```python
# Tabuada/Calculadora v1

oper_aritmetica = input("Digite a operação desejada [ + | - | x | / | ** ]: ")
num_tabuada = float(input("Digite o número da tabuada desejada: "))
contador = 0

if num_tabuada < 0:
    num_formatado = f"({num_tabuada})"
else:
    num_formatado = str(num_tabuada)

if oper_aritmetica == "+" or oper_aritmetica == "-" or oper_aritmetica == "x" or oper_aritmetica == "/" or oper_aritmetica == "**":

    print(f"Tabuada do {num_formatado}:")

    while contador < 11:
        if oper_aritmetica == "+":
            print(f"{num_formatado} + {contador} = {num_tabuada + contador}")
        elif oper_aritmetica == "-":
            print(f"{num_formatado} - {contador} = {num_tabuada - contador}")
        elif oper_aritmetica == "x":
            print(f"{num_formatado} x {contador} = {num_tabuada * contador}")
        elif oper_aritmetica == "**":
            print(f"{num_formatado} ^ {contador} = {num_tabuada ** contador}")
        elif oper_aritmetica == "/":
            if contador == 0:
                pass
            else:
                print(f"{num_formatado} / {contador} = {round(num_tabuada / contador, 1)}")

        contador += 1
else:
    print("Dado(s) inválido(s)! Tente novamente.")
```

```python
# Tabuada/Calculadora v2

oper_aritmetica = input("Digite a operação desejada [ + | - | x | / | ** ]: ")
num_tabuada = float(input("Digite o número da tabuada desejada: "))

num_formatado = f"({num_tabuada})" if num_tabuada < 0 else str(num_tabuada)     # formatação (parênteses para negativos)

operadores_validos = ["+", "-", "x", "/", "**"]

if oper_aritmetica in operadores_validos:
    print(f"Tabuada do {num_formatado}:")

    inicio = 1 if oper_aritmetica == "/" else 0                                 # começa em 1 para evitar divisão por zero

    for contador in range(inicio, 11):
        if oper_aritmetica == "+":
            resultado = num_tabuada + contador
        elif oper_aritmetica == "-":
            resultado = num_tabuada - contador
        elif oper_aritmetica == "x":
            resultado = num_tabuada * contador
        elif oper_aritmetica == "**":
            resultado = num_tabuada ** contador
        elif oper_aritmetica == "/":
            resultado = round(num_tabuada / contador, 1)

        print(f"{num_formatado} {oper_aritmetica} {contador} = {resultado}")
else:
    print("Dado(s) inválido(s)! Tente novamente.")
```
