```python
# Ínicio Tuplas
    # Ordenável, não mutável e indexável

umaTupla = ("maça", "banana", "cereja")
print(umaTupla)

print(len(umaTupla))
```

```python
# Tupla aceita valores duplicados

umaTupla = ("maça", "banana", "cereja", "banana")
print(umaTupla)
```

```python
# Criar Tupla com um item

umaTupla = ("maça",)
print(type(umaTupla))

# Isto não é uma tupla

umaTupla = ("maça")
print(type(umaTupla))

umaLista = ["maça"]
print(type(umaLista))
```

```python
# Diferentes tipos de dado

tupla1 = ("maça", "banana", "cereja")
tupla2 = (1, 5, 7, 9, 3)
tupla3 = (True, False, False)

print(tupla1)
print(tupla2)
print(tupla3)
```

```python
tupla1 = ("arroz", "feijão")
tupla2 = ("e", "carne")

print((tupla1[0],) + tupla2[:])
```

```python
# Adicionar itens a uma Tupla

umaTupla = ("maça", "bacana", "cereja")

y = list(umaTupla)
y.append("laranja")
umaTupla = tuple(y)

print(umaTupla)
```

```python
# Alterar uma Tupla

x = ("maça", "banana", "cereja")
y = list(x)

y[1] = "kiwi"
x = tuple(y)

print(x)
```

```python
x = ("maça", "banana", "cereja")
print(x)
print(type(x))

y = list(x) # construtor
print(y)
print(type(y))
```

```python
# Construtor de Tupla?

naoLista = ("maça")
simLista = list(naoLista)

agrLista = ("maça",)

print(simLista)
print(type(simLista))

print(agrLista)
print(type(agrLista))
```

```python
# Percorrendo um Tupla

umaTupla = ("maça", "banana", "cereja")
for x in umaTupla:
    print(x)
```
