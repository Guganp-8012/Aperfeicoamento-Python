```python
# Lista

lista = [1, 2, 3]
lista_nomes = ["lucas", "jose", "maria"]

# Como apresentar uma lista inteira
print(lista)
print(lista_nomes)
```

```python
# Como setar apenas um valor da lista usando o índice

lista = [1, 2, 3]
x = lista[1]
print(x)


print(lista[0]) # toda lista começa com índice 0
print(lista[1])
print(lista[2])
```

```python
a = int(input("Digita:"))
b = input("Digita:")
lista = [1, a, 3, b]
print(lista)
```

```python
# For em uma lista

lista_nomes = ["lucas", "jose", "maria"]

for nome in lista_nomes:
    print(nome)
    print(lista_nomes)
```

```python
lista_nomes = ["lucas", "jose", "maria"]
print(lista_nomes)

lista_nomes[1] = "josefina"
print(lista_nomes)
```
