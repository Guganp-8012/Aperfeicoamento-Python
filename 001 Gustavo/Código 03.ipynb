```python
# Acumulador de valores

somatorio = 0
contador = 1

while contador <= 3:
  valor = int(input(f"Digite o {contador}º valor:"))
  somatorio = somatorio + valor
  contador = contador + 1

print(f"O somatório total dos valores é de: {somatorio}")
```

```python
# Interrupção da execução do WHILE

somatorio = 0
contador = 1

while contador <= 10:
  valor = int(input(f"Digite o {contador}º valor ou 0 para cancelar:"))
  if valor == 0:
    break
  somatorio = somatorio + valor
  contador = contador + 1

print(f"O somatório total dos valores é de: {somatorio}")
```

```python
somatorio = 0
contador = 1

while contador <= 10:
  valor = int(input("Digite um valor"))
  if valor > 10:
    continue
  somatorio = somatorio + valor
  contador = contador + 1

print(f"O somatório total dos valores é de: {somatorio}")
```
