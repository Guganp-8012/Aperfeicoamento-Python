```python
a = 10
b = 20

# Operadores relacionais

print(a == b)
print(a > b)
print(a < b)
print(a != b)
print(a >= b) # > "ou" = duas verificações, separadas
print(a <= b)
```

```python
media = (float(input("Digite a primeira nota (ex.: 5.5): ")) + float(input("Digite a segunda nota (ex.: 5.5): "))) / 2
print(media, type(media))

if media == 10:
    print("Aprovado e laureado")
elif media >= 9:
    print("Aprovado com honra ao mérito")
elif media >= 7:
    print("Aprovado")
elif media >= 6:
    print("Prova final")
elif media >= 4:
    print("Recuperação")
elif media < 4 and media >= 0:
    print("Reprovado")
else:
    print("Valor Inválido!")
```
