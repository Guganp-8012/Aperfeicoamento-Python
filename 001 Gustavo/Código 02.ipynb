```python
a = 10
b = 20

# Operadores relacionais

print(a == b)     # a "é igual a" b?
print(a > b)      # a "é maior que" b?
print(a < b)      # a "é menor que" b?
print(a != b)     # a "é diferente a" b?
print(a >= b)     # a "é maior ou igual que" b?
print(a <= b)     # a "é menor ou igual que" b?
```

```python
a = 10
b = 20

# Operadores lógicos

print(a == b and 2 * a > b)   # a "é igual a" b "e" 2*a "é maior que" b?
print(a == b or 2 * a > b)    # a "é igual a" b "ou" 2*a "é maior que" b?
print(not a > b)              # a "NÃO é maior que" b?
```

```python
# Estrutura Condicional IF

idade = 17

if idade >= 18:
  print("Pode ir ao show")
```

```python
# Estrutura Condicional IF - ELSE

nota = 8

if nota >= 7:
  print("Aprovado")
else:
  print("Reprovado")
```

```python
# Estrutura Condicional IF - ELSE + INPUT

ano_nascimento = int(input("Digite seu ano de nascimento"))

idade_atual = 2026 - ano_nascimento

if idade_atual >= 18:
    print(f"Você tem {idade_atual} anos, é maior de idade")
else:
    print(f"Você tem {idade_atual} anos, é menor de idade")
```

```python
# Estrutura Condicional IF - ELSE + INPUT + AND

media_final = int(input("Digite sua media final:"))
frequencia = int(input("Digite a porcentagem da sua frequência:"))

if media_final >= 7 and frequencia >= 75:
    print("Você foi aprovado!")
else:
    print("Você foi reprovado!")
```

```python
# Estrutura Condicional IF - ELSE + INPUT + OR

estudante = input("Você é estudante? (sim ou não):")
idoso = input("Você é idoso? (sim ou não):")

if estudante == "sim" or idoso == "sim":
    print("Você está apto a pagar meia-entrada!")
else:
    print("Você deve pagar a entrada por inteiro!")
```

```python
# Estrutura Condicional IF - ELIF - ELSE

media = 7

if media==10:
  print("Aprovado e laureado")
elif media>=9 and media<10:
  print("Aprovado com honra ao mérito")
elif media>=7:
  print("Aprovado")
elif media>=6:
  print("Prova final")
elif media>=4:
  print("Recuperação")
else:
  print("Reprovado")
```
