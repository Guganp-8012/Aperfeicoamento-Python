```python
# Conversão de variáveis

x = str(123)        # Inserindo número porém convertendo para texto
print(type(x))

# aqui defini a variavel y
y = int("123")    # Inserindo texto porém convertendo para número inteiro
print(type(y))

z = float("12.3")    # Inserindo texto porém convertendo para número decimais(fracionados)
print(type(z))


print("tenho " + str(20 + 2) + " anos")
print("tenho", 20 + 2, "anos")
print(f"tenho {20 + 2} anos")
```

```python
# Cálculo matemático

soma = 1 + 1
print(f"Resultado soma: {soma}")

subtracao = 2 - 1
print(f"Resultado subtração: {subtracao}")

multiplicacao = 2 * 7
print(f"Resultado multiplicação: {multiplicacao}")

divisao = 10 / 3
print(f"Resultado divisão: {divisao}")

divisao_inteira = 10 // 3                    # Resultado da divisão sem valores decimais
print(f"Resultado divisão inteira: {divisao_inteira}")

modulo = 11 % 2                              # Resto da divisão
print(f"Resultado módulo: {modulo}")

potencia = 2 ** 3                            # 2x2x2
print(f"Resultado potência: {potencia}")
```
